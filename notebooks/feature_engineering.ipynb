```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/data.csv')
df.head()
```

```python
df.info()
```

```python
df.rename(
    columns={
        "air_store_id": "store_id",
        "visit_date": "date",
        "day_of_week": "dow",
        "air_genre_name": "store_genre",
        "air_area_name": "area"
    },
    inplace=True
)
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')
df.head()
```

```python
df['month'] = df['date'].dt.month
df.head()
```

```python
df['weekend_flg'] = df['dow'].isin(['Saturday', 'Sunday']).astype(int)
df.head()
```

```python
df.sort_values(['store_id', 'date'], inplace=True)
df.head()
```

```python
df.reset_index(drop=True, inplace=True)
df.head()
```

```python
for lag_days in [1]:
    temp = df[['store_id', 'date', 'visitors']].copy()
    temp['date'] = temp['date'] + pd.Timedelta(days=lag_days)
    temp.rename(columns={'visitors': f'visitors_lag_{lag_days}'},inplace=True)
    df = pd.merge(df, temp, on=['store_id', 'date'], how='left')

df.head(7)
```

```python
df['visitors_last_day'] = df.groupby('store_id')['visitors'].shift(1)
df.head()
```

```python
roll = (
    df.set_index('date')
      .groupby('store_id')['visitors']
      .rolling('7D', closed='left')
      .mean()
      .reset_index()
      .rename(columns={'visitors': 'visitors_mean_roll_7'})
)

df = pd.merge(df, roll, on=['store_id', 'date'], how='left')
df.head(7)
```
